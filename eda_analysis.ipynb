```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
# Task 1
iris = pd.read_csv(r'M:\codaveda\1) iris.csv')

iris.head(100)

```

```python
iris.duplicated().sum() # check for duplicates
print(iris.isna().sum()) # check for missing values
iris.describe() # statistical summary of the dataset
```

```python
# Task 2 ==> Exploratory Data Analysis (EDA)
iris_data_analysis = iris.groupby('species')['petal_length'].mean().reset_index()
iris_data_analysis

```

```python
plt.figure(figsize=(10, 6))
plt.bar(iris_data_analysis['species'], iris_data_analysis['petal_length'], color=['blue', 'orange', 'green'])
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.title('Average Petal Length by Species')
plt.show()
```

```python
#Average petal length increases significantly from Setosa to Versicolor, reaching its peak in Virginica.
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='petal_length', data=iris)
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.title('Box Plot of Petal Length by Species')
plt.show()
```

```python
# Setosa has the smallest petal length but shows several outliers (dots below and above the box). Versicolor also has one outlier at the bottom.
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', data=iris)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.legend(title='Species')
plt.show()
```

```python
# There is a clear positive correlation between petal length and width; Setosa is easily separable from the other two species.
```
